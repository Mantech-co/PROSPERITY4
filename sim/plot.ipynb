```python
from bokeh.plotting import figure, show, output_notebook, column
from bokeh.models import ColumnDataSource, HoverTool
import pandas as pd
```

```python
def bokeh_plot(x, y, title="Plot", x_label="x", y_label="y", x_range=None):
    output_notebook()

    df = pd.DataFrame({"x": x, "y": y}).dropna()
    source = ColumnDataSource(df)

    fig_kwargs = dict(
        title=title,
        x_axis_label=x_label,
        y_axis_label=y_label,
        tools="pan,wheel_zoom,box_zoom,reset,save",
        width=1200,
        height=300
    )

    if x_range is not None:
        fig_kwargs["x_range"] = x_range
    
    p = figure(**fig_kwargs)

    p.line("x", "y", source=source)
    p.scatter("x", "y", source=source, size=3)
    p.add_tools(HoverTool(tooltips=[("x", "@x"), ("y", "@y")]))

    return p
```

```python
df1 = pd.read_csv("TUTORIAL_ROUND_1/prices_round_0_day_-1.csv", sep=";")

df1 = df1[df1["product"] == "TOMATOES"]

df1["mid_price"] = pd.to_numeric(df1["mid_price"], errors="coerce")
df1 = df1.dropna(subset=["mid_price"])

df2 = df1.copy()
df3 = df1.copy()

df2["ask_price_1"] = pd.to_numeric(df2["ask_price_1"], errors="coerce")
df2 = df2.dropna(subset=["ask_price_1"])

df3["bid_price_1"] = pd.to_numeric(df3["bid_price_1"], errors="coerce")
df3 = df3.dropna(subset=["bid_price_1"])

df4 = pd.read_csv("TUTORIAL_ROUND_1/trades_round_0_day_-1.csv", sep=';')
df4 = df4[df4["symbol"] == "TOMATOES"]
df4 = df4[df4["quantity"] == 2]

df4["price"] = pd.to_numeric(df4["price"], errors="coerce") 
df4 = df4.dropna(subset=["price"])

from bokeh.models import Range1d

xmin = min(df1["timestamp"].min(), df2["timestamp"].min())
xmax = max(df1["timestamp"].max(), df2["timestamp"].max())

shared_range = Range1d(xmin, xmax)
```

```python
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource
import pandas as pd

output_notebook()

# merge properly
df = pd.merge(df1, df2, on="timestamp", how="outer")
df = pd.merge(df, df3, on="timestamp", how="outer")
df = pd.merge(df, df4, on="timestamp", how="outer")

df = df.sort_values("timestamp")

source = ColumnDataSource(df)

p = figure(
    title="TOMATOES: Trades vs Mid Price",
    x_axis_label="Time",
    y_axis_label="Price",
    width=1200,
    height=400,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

# line 1
p.line("timestamp", "mid_price", source=source, line_width=2, color="blue", legend_label="Mid price")

p.line("timestamp", "ask_price_1", source=source, line_width=2, color="red", legend_label="Best ask price")

p.line("timestamp", "bid_price_1", source=source, line_width=2, color="green", legend_label="Best bid price")

p.scatter("timestamp", "price", source=source, size=4, color="black", legend_label="Trades")

# enable legend toggle
p.legend.click_policy = "hide"

show(p)
```
